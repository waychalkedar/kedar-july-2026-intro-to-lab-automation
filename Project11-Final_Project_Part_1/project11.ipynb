```python
import serial
import sys
```

```python
arduino = serial.Serial('COM4', 9600, timeout=1)

try:
    while True:
        data = arduino.readline()
        if not data:
            continue
        data = data.rstrip(b'\r\n')
        parts = data.split(b' ')
        angle = parts[0][0]
        buzzer = bool(parts[1][0])

except KeyboardInterrupt:
    pass
except Exception:
    import traceback
    traceback.print_exc()
finally:
    arduino.close()
    sys.exit()
```

```python
arduino = serial.Serial('COM4', 9600, timeout=1)

try:
    while True:
        data = arduino.read()
        if data:
            print(data)
            print(int.from_bytes(data, "little"))
            print("")

except KeyboardInterrupt:
    pass
finally:
    arduino.close()
    sys.exit()
```
